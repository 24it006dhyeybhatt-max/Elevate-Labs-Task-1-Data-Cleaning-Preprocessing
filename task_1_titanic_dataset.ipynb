# **2.Handle missing values using mean/median/imputation.**

```python
#Handle Missing Age Values Using Median

df["Age"] = df["Age"].fillna(df["Age"].median())

print("Missing Age Values After Imputation:")

print(df["Age"].isnull().sum())
```

```python
#Handle Missing Embarked Values Using Mode

df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])

print("Missing Embarked Values After Imputation:")

print(df["Embarked"].isnull().sum())
```

```python
#Handle Missing Cabin Values Using Imputation

df["Cabin"] = df["Cabin"].fillna("Unknown")

print("Missing Cabin Values After Imputation:")

print(df["Cabin"].isnull().sum())
```

```python
#Verify Missing Values After Cleaning

print("Missing Values After Cleaning:")

print(df.isnull().sum())

print("\nTotal Missing Values:", df.isnull().sum().sum())
```

```python
#Visualize Missing Values After Cleaning

plt.figure(figsize=(12, 6))

sns.heatmap(df.isnull(), cbar=False, cmap="viridis")

plt.title("Missing Values After Data Cleaning")

plt.xlabel("Columns")

plt.ylabel("Passenger Records")

plt.show()
```

# **3.Convert categorical features into numerical using encoding.**

```python
#Import LabelEncoder

from sklearn.preprocessing import LabelEncoder
```

```python
#Apply Label Encoding to Sex Column

le = LabelEncoder()

df["Sex_Encoded"] = le.fit_transform(df["Sex"])

print(df[["Sex", "Sex_Encoded"]].head())
```

```python
#Apply One-Hot Encoding to Embarked Column if it exists

if 'Embarked' in df.columns:
    df = pd.get_dummies(
        df,
        columns=["Embarked"],
        dtype=int
    )
else:
    print("'Embarked' column already encoded or not found.")

print(df.head())
```

```python
#Display Encoded Categorical Features

print("Encoded Dataset:")

print(df[["Sex", "Sex_Encoded"]].head())

print("\nOne-Hot Encoded Columns:")

print(df[["Embarked_C", "Embarked_Q", "Embarked_S"]].head())
```

# **5.Visualize outliers using boxplots and remove them.**

```python
#Visualize Outliers Using Boxplots

plt.figure(figsize=(12, 6))
sns.boxplot(data=df[["Age", "Fare"]])
plt.title("Boxplot of Age and Fare Before Removing Outliers")
plt.xlabel("Numerical Features")
plt.ylabel("Values")
plt.show()
```

```python
#Detect Outliers Using IQR Method

Q1 = df[["Age", "Fare"]].quantile(0.25)
Q3 = df[["Age", "Fare"]].quantile(0.75)
IQR = Q3 - Q1
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR
print("Lower Limits:")
print(lower_limit)
print("\nUpper Limits:")
print(upper_limit)
```

```python
#Count Outliers in Numerical Features

for column in ["Age", "Fare"]:

    outliers = df[
        (df[column] < lower_limit[column]) |
        (df[column] > upper_limit[column])
    ]

    print(column, "Outliers:", len(outliers))
```

```python
#Remove Outliers Using IQR Method

df_cleaned = df[
    (df["Age"] >= lower_limit["Age"]) &
    (df["Age"] <= upper_limit["Age"]) &
    (df["Fare"] >= lower_limit["Fare"]) &
    (df["Fare"] <= upper_limit["Fare"])
]
print("Original Dataset Shape:", df.shape)
print("Cleaned Dataset Shape:", df_cleaned.shape)
print("Total Rows Removed:", len(df) - len(df_cleaned))
```

```python
#Visualize Outliers After Removal

plt.figure(figsize=(12, 6))
sns.boxplot(data=df_cleaned[["Age", "Fare"]])
plt.title("Boxplot After Removing Outliers")
plt.xlabel("Numerical Features")
plt.ylabel("Values")
plt.show()
```

```python
#Compare Dataset Before and After Outlier Removal

print("Before Outlier Removal:")
print("Rows:", df.shape[0])

print("\nAfter Outlier Removal:")
print("Rows:", df_cleaned.shape[0])

print("\nTotal Rows Removed:")
print(df.shape[0] - df_cleaned.shape[0])
```

```python
#Save Final Cleaned Dataset

df_cleaned.to_csv("Titanic_Cleaned_Dataset.csv", index=False)

print("Final Cleaned Dataset Saved Successfully!")
```
